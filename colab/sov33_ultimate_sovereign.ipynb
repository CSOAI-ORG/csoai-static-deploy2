```python
# Load model
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "nicholasgriffintn/sov33-ultimate-sovereign"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype=torch.float16, device_map="auto")

print(f"Model loaded: {model_name}")
print(f"Device: {model.device}")
```

```python
# Chat function
def chat(message, max_tokens=256):
    inputs = tokenizer(message, return_tensors="pt").to(model.device)
    with torch.no_grad():
        outputs = model.generate(**inputs, max_new_tokens=max_tokens, temperature=0, do_sample=False)
    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return response[len(message):].strip()

# Test
print(chat("What is the EU AI Act Article 50?"))
```

```python
# Interactive chat
while True:
    message = input("You: ")
    if message.lower() in ["quit", "exit", "q"]:
        break
    response = chat(message)
    print(f"SOV33: {response}")
```
